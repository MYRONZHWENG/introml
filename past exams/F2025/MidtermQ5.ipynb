## Solution

```python
m = np.random.randint(2, 10)   
p = np.random.randint(2, 10) 
n = np.random.randint(2, 10)   

Q = np.random.randint(-20, 21, size=(m, p))
X = np.random.randint(-20, 21, size=(n, p))

print(f"\nQ: shape:({m},{p})")
print(Q)

print(f"\nX: shape:({n},{p})")
print(X)
```

```python
dist = np.sum((Q[:, None, :] - X[None, :, :])**2, axis=2) # (m,n), dist[i,j]:=distance b/w Q[i] and X[j].

print(f"\ndist: shape:{dist.shape}")
print(dist)
```

```python
minq_ind = np.argmin(dist, axis=0) #(n,), row index of row vector of Q that has smallest distance from ith row vector of X

print("\nminq_ind:")
print(minq_ind)
```

```python
type(minq_ind) #np.min returns an nparray
```

```python
qhat = Q[minq_ind] #(n.p), qhat[i] = Q[minq_ind[i]]

print(f"\nQ: shape:({m},{p})")
print(Q)

print(f"\nqhat: shape{qhat.shape}")
print(qhat)

print(f"\nX: shape:({n},{p})")
print(X)
```

```python
col_ind = np.arange(dist.shape[1]) # (n,)
dhat = dist[minq_ind, col_ind] # (n,), select minimum distances

print(f"\ndhat: shape{dhat.shape}")
print(dhat.reshape(-1,1))
```

### part(b)

```python
A = np.hstack((np.sqrt(dhat).reshape(-1,1),qhat)) # (n,p+1)

print(f"\nA: shape{A.shape}")
print(A)
```

```python
y = np.random.randint(-20, 21, (n)) #(n,)

print(f"\ny: shape{y.reshape(-1,1).shape}")
print(y.reshape(-1,1))
```

```python
beta = np.linalg.lstsq(A,y) # (p+1,)

desc = ['lstsq soln','residuals','rank','singular values']
for i in range(len(beta)):
    print("\n" + desc[i] + ":")
    print(beta[i])
```

```python
alpha = beta[0][0]
w = beta[0][1:]

print(f"\nalpha: shape{alpha.shape}")
print(alpha)

print(f"\nw: shape{w.reshape(-1,1).shape}")
print(w.reshape(-1,1))
```
