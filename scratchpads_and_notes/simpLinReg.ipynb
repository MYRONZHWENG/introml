```python
import numpy as np
from sklearn import linear_model
from sklearn.metrics import r2_score
```

```python
A = np.array([[0,1,0,2],[0,1,0,3],[1,0,-1,0],[1,0,0,0]])
y = np.array([19,26,7,2])#.reshape(-1,1)
```

```python
linreg = linear_model.LinearRegression(fit_intercept=False)
linreg.fit(A,y)
```

```python
col_ones = np.ones((4,1))
padded_A = np.hstack((col_ones,A))
```

```python
padded_A
```

```python
aTa = padded_A.T.dot(padded_A)
```

```python
aTa # aTa is not actually invertible
```

```python
np.linalg.det(aTa)
```

```python
from sympy import Matrix
aTa_matrix = Matrix(aTa)
```

```python
rref_form, pivot_columns = aTa_matrix.rref()
print(rref_form)
print('pivot columns: ', pivot_columns)
```

```python
print(aTa_matrix.det()) # Matrix from sympy package does determinant correctly, not numpy.linalg
```

**Note**

Padding your design matrix with a column of ones is not a requirement, nor is removing the mean from it. You do this when you want to include an intercept in your linear model.

For the following two methods:
1) padding design matrix with a colum of ones
2) subtracting the mean of each column of the design matrix from each element and doing the same with the output matrix

Both methods give the same coefficient results except for the intercept because the intercept is not calculated for method `2`. (I don't know mathematical proof for yet)

If the calculated coefficient matrix from method `1` has dimensions `n x 1`, that of method `2` is `(n-1) x 1`, excluding the intercept at the first element (row 1 column 1)


$$\vec{β_{padded}} = \begin{pmatrix} \beta_0 \\ \beta_1 \\ \vdots \\ \beta_p \end{pmatrix} ,  \vec{β_{noMean}} = \begin{pmatrix} \beta_1 \\ \beta_2 \\ \vdots \\ \beta_p \end{pmatrix}$$
