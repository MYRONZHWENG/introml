## Reshaping:

```python
a = np.arange(5)
print(a)
print(a.shape)
print(a[None,:])
print((a[None,:]).shape)
print(a[:,None])
print((a[:,None]).shape)
```

```python
print(a.reshape(5,1))
print((a.reshape(5,1)).shape)
```

```python
b = a[:,None]
print(b)
print(b.shape)
b = b[None,:]
print(b)
print(b.shape)
```

**Note** 

Every time you use [None,:] or [:,None], you are adding a new (higher) dimension to your original array

But not if you are using reshape()

```python
print('a = ', a)
c = a.reshape(5,1)
print(c)
print(c.shape)
c = c.reshape(1,5)
print(c)
print(c.shape)
```

### Testing broadcasting with redundant dimensions (using xts and yts)

```python
xts.shape, yts.shape
```

```python
xts = (xts[None,:])[:,None] # want to make into column vector
yts = (yts[None,:])[None,:] # want to make into row vector

xts.shape,yts.shape
```

```python
zts = xts * yts
zts # we still get a vector inside an array that is inside an array, not a 2D array
```

**Note**

When we do `xts[None,:]`, assuming `xts` is of shape `(10,)` at start, we make `xts` of shape `(1,10)` as expected. But, when we run `[:,None]` on `xts` immediatedly afterwards, `xts` is now a `(1,1,10)` array because we are treating **the entire `(1,10)` array as one element** and putting it in a single dimension array with one column.

Use `reshape()` instead to broadcast properly

```python
print(xts)
```

```python
print(yts)
```

### trying to get (1,10,1)

```python
v = np.arange(0,10)
v.shape
```

```python
v = v[:,None][None,:]
print(v)
print(v.shape)
```

**Note**

`(1,10)` is technically a `(10,)` inside an array with one row and one column because we are treating the entire `(10,)` as one element

`:` means get every element, but the only element there is is the `(10,)` array

```python
t = np.arange(0,10)
print(t)
print(t.shape)
t = t[None,:] # get every element and put it in a new row
print(t)
print(t.shape)
```

## Using reshape:


In NumPy, reshape(-1, 1) is a method used to change the shape of an array.

    -1: This acts as a wildcard. When used in reshape, it tells NumPy to automatically calculate the size of that dimension based on the total number of elements in the array and the sizes of the other specified dimensions.
    1: This explicitly sets the size of the second dimension (columns) to 1. 

Therefore, array.reshape(-1, 1) reshapes a 1-D array into a 2-D array with a single column. The number of rows will be automatically determined by NumPy to accommodate all elements from the original array.

```python
s = np.arange(0,10)
print(s.reshape(-1,1))
print(s.reshape(-1,1).ravel())
```
